This notebook used to contain a full duplicate copy of the simulation engine

For a non-interactive run, use the CLI instead: `iidsim run --dataset <name> --corridor <corridor>` (see `iidsim run --help` and `docs/simulation-engine.md`).

```python
from iidsim.engine import run_simulation
from iidsim.schedules import available_corridors

available_corridors()
```

```python
result = run_simulation(
    corridor_dataset="p_g_sprd_vzm_2days",
    network_section="sprd_vzm",
    output_dir="output_files",
)
result["excel_filename"], result["chart_filename"], result["animator_filename"]
```

```python
# Inspect one train's planned vs simulated schedule
for tr in result["trains"][:3]:
    print(tr.train_id, tr.calc_tr_stats())
```
